#Importing the Dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn import metrics
```

```python
boston_df = pd.read_csv('/content/boston.csv')
```

```python
print(boston_df)
```

```python
boston_df.head()
```

```python
boston_df.describe()
```

```python
boston_df.shape
```

```python
boston_df.isnull().sum() #checking the missing values
```

#checking the correlation

```python
bos_corr = boston_df.corr()
```

```python
# construction of heatmap to depict the correlation
plt.figure(figsize=(8,8))
sns.heatmap(bos_corr,cbar=True,square=True,fmt='.1f',annot=True,annot_kws={'size':8},cmap='Blues')
```

#Splitting the data and Target

```python
X = boston_df.drop(['Price'],axis=1)
Y = boston_df['Price']
```

```python
print(X)
print(Y)
```

#Splitting the Data into Training and Testing Data

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=2)
print(X.shape,X_train.shape,X_test.shape)
```

#XGBoostRegressor

```python
#loading the XGBoost
model = XGBRegressor()
```

```python
model.fit(X_train,Y_train) #training the model with X_train and Y_train
```

#Evaluation

```python
#accuracy score cannot be defined for regression models
#Prediction on training data
```

```python
training_data_prediction = model.predict(X_train)
#X_train consists of training values and Y_train consists of real values
```

```python
print(training_data_prediction)
```

```python
# R squared error
score_1 = metrics.r2_score(Y_train,training_data_prediction)
```

```python
print('R Square error value is : ',score_1)
```

```python
#MeanAbsoluteError
score_2 = metrics.mean_absolute_error(Y_train,training_data_prediction)
```

```python
print('Mean Absolute error value is : ',score_2)
```

#Visulaizing Actual prices and predicted values

```python
plt.scatter(Y_train,training_data_prediction)
plt.xlabel("Actual Prices")
plt.ylabel("Predicted Prices")
plt.title("Actual Prices Vs Predicted Values")
plt.show()
```

#Prediction on Test Data

```python
test_data_prediction = model.predict(X_test)
```

```python
print(test_data_prediction)
```

```python
score_3 = metrics.r2_score(Y_test,test_data_prediction)
print('R Square error value is : ',score_3)

score_4 = metrics.mean_absolute_error(Y_test,test_data_prediction)
print('Mean Absolute error value is : ',score_4)
```
