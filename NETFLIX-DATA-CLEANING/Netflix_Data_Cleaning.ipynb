## **EXPLORATORY DATA ANALYSIS**

## IMPORTING DATA VISUALIZATION LIBRARIES

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### CONTENT TYPE DISTRIBUTION

```python
types = df['type'].value_counts()
print(types)
```

```python

plt.figure(figsize=(8,8))
sns.barplot(x=types.index, y=types.values, palette='Set1' )
plt.title('CONTENT TYPE DISTRIBUTION')
plt.xlabel('CONTENT TYPES')
plt.ylabel('COUNT')
plt.show()
```

### GENRE DISTRIBUTION

```python
## SPLITS GENRES IN EXISTING COLUMN AND ADD IT THOSE INTO NEW 'GENRE' COLUMN
df['genres'] = df['listed_in'].apply(lambda x: x.split(', '))

## CONVERTS LISTS OF GENRES INTO ONE LIST
all_genres = sum(df['genres'], [])

## CONVERTS LIST INTO PANDAS SERIES AND COUNTS THEIR REPETITION
genre_counts = pd.Series(all_genres).value_counts().head(10)

## BAR PLOT
plt.figure(figsize=(8,6))
sns.barplot(x = genre_counts.values, y = genre_counts.index ,palette='pastel' )
plt.title("GENRE DISTRIBUTION")
plt.xlabel("COUNT")
plt.ylabel("GENRES")
plt.show()
```
