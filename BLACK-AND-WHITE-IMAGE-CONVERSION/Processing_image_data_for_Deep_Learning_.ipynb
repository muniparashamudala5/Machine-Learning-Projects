```python
# Resizing the image using PIL
from PIL import Image
img = Image.open("/content/Parash's Image.jpg")
img_resized = img.resize((200,200))
```

# **CONVERTING RGB TO GRAYSCALE**

```python
img = cv2.imread("/content/Parash's Image.jpg")
```

```python
type(img)
```

```python
grayscaled_img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
```

```python
type(grayscaled_img)
```

```python
from google.colab.patches import cv2_imshow
cv2_imshow(grayscaled_img)
```

```python
img_gray_plot = plt.imshow(grayscaled_img)
plt.show()
```
